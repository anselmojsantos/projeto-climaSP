```python
import geopandas as gpd

zonas = gpd.read_file(os.path.join(RAIZ, "zonas_sp.geojson"))

# Expande as zonas em 100 m (EPSG:31983) para incluir Cidade Tiradentes 2 (~34 m fora da Leste)
zonas["geometry"] = zonas.to_crs("EPSG:31983").geometry.buffer(40).to_crs("EPSG:4326").geometry

gdf = gpd.GeoDataFrame(
    df_registro,
    geometry=gpd.points_from_xy(df_registro["longitude"], df_registro["latitude"]),
    crs="EPSG:4326",
)
gdf = gpd.sjoin(gdf, zonas, how="left", predicate="within")
gdf["zona"] = gdf["zona"].fillna("Fora de SP")

df_station_sp = gdf.drop(columns=["geometry", "index_right"]).copy()

df_plot = df_station_sp.merge(
    df_registro[["id_reading", "description"]],
    on="id_reading",
    how="left"
)

df_plot = df_plot[
    df_plot["zona"].isin(["Sul", "Leste", "Norte", "Oeste", "Centro"])
].copy()
print(df_plot["zona"].value_counts())
```

```python
# Agrupando os dados do CEMADEN por Ano, Trimestre e Região
# (cria as colunas exigidas a partir das colunas reais do DataFrame df_plot)
df_cemaden_agrupado = (
    df_plot.assign(
        Ano=pd.to_datetime(df_plot["observation_date"]).dt.year,
        Trimestre=pd.to_datetime(df_plot["observation_date"]).dt.quarter,
        Regiao=df_plot["zona"],
        Valor=df_plot["value"],
    )
    .groupby(["Ano", "Trimestre", "Regiao"], as_index=False)
    .agg(
        Media_Chuva_240m=("Valor", "mean"),
        Total_Chuva=("Valor", "sum"),
        Eventos_Extremos=("Valor", lambda x: (x > 10).sum()),
    )
)

df_cemaden_agrupado.head()
```

```python
# Mapa das estacoes com pinos coloridos por zona (Icon)
import folium

cores = {
    "Centro": "gray",
    "Norte": "darkblue",
    "Sul": "darkgreen",
    "Leste": "red",
    "Oeste": "darkred",
}

m = folium.Map(
    location=[df_station_sp["latitude"].mean(), df_station_sp["longitude"].mean()],
    zoom_start=12,
)

folium.TileLayer(
    tiles="https://mt1.google.com/vt/lyrs=m&x={x}&y={y}&z={z}",
    attr="Google",
    name="Google Maps",
).add_to(m)

for _, r in df_plot.iterrows():
    cor = cores.get(r["zona"], "gray")
    nome = r.get("nome", r.get("name", "Estação"))
    folium.Marker(
        location=[r["latitude"], r["longitude"]],
        popup=f"{nome} - {r['zona']}",
        tooltip=nome,
        icon=folium.Icon(color=cor)
    ).add_to(m)

CAMINHO = os.path.join(RAIZ, "stations_map.html")
m.save(CAMINHO)
print("Mapa salvo em:", CAMINHO)
m
```
